```python
"""
Импорт библиотек
"""
import invest_toolkit as it
import pandas as pd
```

```python
it.log.init('DATA')
```

```python
"""
Длинная таблица со всеми даннныыми по всем акциям MOEX
"""
df = it.all_stock_info(
    free_dloat_path=r'../support_files/20251206-free-float.xlsx',
    ir_path=r'../support_files/ir_rating.xlsx',
    sl_stock_folder=r'../support_files/scrapper_reports',
)
display(df.head())
df.to_csv(r'../.output/all_stock_info.csv', index=False)
```

```python
"""
Расчет рейтинга для всех акций
"""
rating = it.rating_df(df, 7)
```

```python
"""
Подготовка таблицы рыйтинга для yandex datalens
"""
names_and_url = df[df['indicator'].isin([
    'name',
    'url'])]
rating_to_datalens = pd.concat([names_and_url,rating])
rating_to_datalens = rating_to_datalens.pivot(
    index=['ticker'],
    columns='indicator',
    values='value'
    )
display(rating_to_datalens.head())
rating_to_datalens.to_csv(r'../.output/datalens_rating.csv')
```

```python
"""
Подготовка данных с показателями для yandex datalens
"""
indicators_to_datalens = df[df['indicator'].isin([
    'Див.выплата, млрд руб', 
    'Чистая прибыль, млрд руб', 
    'Чистый долг, млрд руб', 
    'Долг/EBITDA',])]

indicators_to_datalens.to_csv(r'../.output/datalens_indicators.csv')
```
